```python
import ast  #abstract syntax tree-just used for seeing the string representatin
            #of python objects -not used in project

import pandas as pd

# Convert category dictionaries into individual labels
#converte and return the categories into true/false -returns only if it is true

def get_categories(category_dict):
    return [category for category, value in category_dict.items() if value]

#We extract the active categories from the category dictionaries using get_categories().
#To find all unique categories, we can use explode() to separate the lists and unique() to remove duplicates.
#In our experiment, the MultiLabelBinarizer identified 14 categories.

train_df["labels"] = train_df["category"].apply(get_categories)
test_df["labels"] = test_df["category"].apply(get_categories)

# Combine prompt + response
train_df["text"] = train_df["prompt"] + " " + train_df["response"]
test_df["text"] = test_df["prompt"] + " " + test_df["response"]

print(train_df[["text", "labels"]].head())
```

This cell prepares the data for model training. It defines a helper function `get_categories` that extracts 'True' categories from the `category` dictionary. It then applies this function to create a new 'labels' column for both `train_df` and `test_df`, containing lists of active categories. Finally, it concatenates the 'prompt' and 'response' columns into a new 'text' column for both DataFrames, which will be used as the input features for the model. It then prints the head of the 'text' and 'labels' columns.

```python
from sklearn.preprocessing import MultiLabelBinarizer   

#MultiLabelBinarizer.
#Convert multiple text labels into a binary numerical representation.
#It converts each list of labels into 0 and 1.
#For example, suppose the learned category order is:

mlb = MultiLabelBinarizer()
#this is pre one hot encoding of labels and cotegories

y_train = mlb.fit_transform(train_df["labels"])
y_test = mlb.transform(test_df["labels"])

print("Number of categories:", len(mlb.classes_))
print(mlb.classes_)
```

This cell utilizes `MultiLabelBinarizer` from `sklearn.preprocessing` to convert the list of category labels into a binary (one-hot encoded) numerical representation. It fits the binarizer on the `train_df['labels']` to learn all unique categories and then transforms both training (`y_train`) and testing (`y_test`) labels into this binary format. It also prints the total number of categories found and the names of these categories.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

#Term Frequency-Inverse Document Frequency) is a statistical metric used in
#natural language processing to evaluate how important
#a word is to a document within a larger collection or corpus

vectorizer = TfidfVectorizer(
    max_features=30000,
    stop_words="english",
    ngram_range=(1, 2)
)

X_train = vectorizer.fit_transform(train_df["text"])
X_test = vectorizer.transform(test_df["text"])

print(X_train.shape)
```

This cell uses `TfidfVectorizer` from `sklearn.feature_extraction.text` to convert the text data ('prompt' + 'response') into numerical features. It initializes the vectorizer with `max_features` to limit the vocabulary size, `stop_words='english'` to remove common English stop words, and `ngram_range=(1, 2)` to consider both single words (unigrams) and two-word phrases (bigrams). It then fits the vectorizer on the `train_df['text']` and transforms both `train_df['text']` and `test_df['text']` into TF-IDF matrices (`X_train` and `X_test`). Finally, it prints the shape of `X_train`, which represents the number of samples and features.

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression

#One classifier is trained for each category,
 #treating that category as "Yes" and all other categories as "No".
 
model = OneVsRestClassifier(
    LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    )
)

model.fit(X_train, y_train)

print("MODEL TRAINED SUCCESSFULLY!")
```

This cell trains a multi-label classification model using `OneVsRestClassifier` and `LogisticRegression`. `OneVsRestClassifier` trains a separate binary classifier for each label, effectively handling the multi-label nature of the problem. The `LogisticRegression` model is configured with `max_iter=1000` for convergence and `class_weight='balanced'` to handle potential class imbalance. The model is then trained on `X_train` (TF-IDF features) and `y_train` (binary labels). A success message is printed after training.

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test)

print(classification_report(
    y_test,
    y_pred,
    target_names=mlb.classes_,
    zero_division=0
))
```

This cell evaluates the performance of the trained model on the test data. It first uses the `model.predict(X_test)` to get predictions (`y_pred`) for the test set. Then, it prints a detailed `classification_report` using `sklearn.metrics`, comparing the true labels (`y_test`) with the predicted labels (`y_pred`). The report includes precision, recall, f1-score, and support for each category, as well as micro, macro, and weighted averages, providing a comprehensive view of the model's performance.

```python
#vectorizer is the TF-IDF vectorizer you trained earlier.

#The [text] is used because transform() expects
#a list of documents, even if you're predicting only one text.

def predict_text(text):

    # Convert input into TF-IDF
    text_vector = vectorizer.transform([text])

    # Predict categories
    prediction = model.predict(text_vector)

    # Convert prediction back to category names
    categories = mlb.inverse_transform(prediction)

    if len(categories[0]) == 0:
        return ["Safe"]

    return list(categories[0])
```

This cell defines a helper function `predict_text` that takes a single text input, converts it into a TF-IDF vector using the previously trained `vectorizer`, and then uses the `model` to predict the categories. It converts the numerical predictions back into human-readable category names using `mlb.inverse_transform`. If no categories are detected, it returns 'Safe'.

```python

import re

def explain_prediction(labels):
    explanations = []

    # Normalize predicted labels.
    labels_list = [
        str(label).strip().lower()
        for label in (labels or [])
        if label is not None and str(label).strip()
    ]

    # Exact mapping to the 14 categories found in your notebook's mlb.classes_.
    # All actual model categories are explicitly mapped here.
    CATEGORY_EXPLANATIONS = {
        "animal_abuse": (
            "The content may involve abuse, cruelty, or harm toward animals."
        ),

        "child_abuse": (
            "The content may involve abuse, exploitation, or harm toward children."
        ),

        "controversial_topics,politics": (
            "The content relates to controversial topics or political issues."
        ),

        "discrimination,stereotype,injustice": (
            "The text may contain discriminatory, stereotypical, "
            "or unjust language toward a person or group."
        ),

        "drug_abuse,weapons,banned_substance": (
            "The content may involve drug abuse, weapons, or prohibited substances."
        ),

        "financial_crime,property_crime,theft": (
            "The content may involve financial crime, property crime, "
            "theft, or potentially illegal activity."
        ),

        "hate_speech,offensive_language": (
            "The text contains offensive or potentially hateful language."
        ),

        "misinformation_regarding_ethics,laws_and_safety": (
            "The content may contain misinformation concerning ethics, "
            "laws, or safety."
        ),

        "non_violent_unethical_behavior": (
            "The content may describe unethical behavior that does not "
            "necessarily involve physical violence."
        ),

        "privacy_violation": (
            "The text may involve sensitive personal or private information."
        ),

        "self_harm": (
            "The text contains content related to self-harm or suicide."
        ),

        "sexually_explicit,adult_content": (
            "The text contains sexual or potentially explicit adult content."
        ),

        "terrorism,organized_crime": (
            "The content may involve terrorism or organized criminal activity."
        ),

        "violence,aiding_and_abetting,incitement": (
            "The content may involve violence, assistance in harmful acts, "
            "or encouragement of violence."
        )
    }

    # Preserve the original keywords as aliases for legacy category names.
    # These are not treated as additional model output categories.
    LEGACY_KEYWORDS = {
        "discrimination": [
            "discrimination", "stereotype", "injustice"
        ],
        "hate_speech": [
            "hate_speech", "offensive_language"
        ],
        "threat": [
            "threat", "threatening"
        ],
        "violence": [
            "violence", "violent"
        ],
        "harassment": [
            "harassment", "harassing"
        ],
        "abuse": [
            "abuse", "abusive"
        ],
        "toxicity": [
            "toxic", "toxicity"
        ],
        "sexual_content": [
            "sexual", "sexually", "explicit"
        ],
        "self_harm": [
            "self_harm", "self-harm", "suicide"
        ],
        "privacy": [
            "privacy", "private_information", "personal_information"
        ],
        "prompt_injection": [
            "prompt_injection", "injection"
        ],
        "cyber_threat": [
            "cyber", "cyber_threat", "cybersecurity"
        ],
        "illegal_activity": [
            "illegal", "illegal_activity"
        ],
        "non_violent": [
            "non_violent", "non-violent"
        ],
        "safe": [
            "safe"
        ]
    }

    # Match complete tokens, treating underscores as part of a token.
    # This prevents 'violent' from matching inside 'non_violent'.
    def contains_keyword(text, keyword):
        pattern = (
            r"(?<![a-z0-9_])"
            + re.escape(keyword.lower())
            + r"(?![a-z0-9_])"
        )
        return re.search(pattern, text) is not None

    # First, use exact matches against the actual model taxonomy.
    matched_labels = set()

    for label in labels_list:
        if label == "safe":
            continue

        if label in CATEGORY_EXPLANATIONS:
            explanations.append(CATEGORY_EXPLANATIONS[label])
            matched_labels.add(label)

    # Support legacy labels without allowing them to override
    # the meaning of an actual model category.
    for label in labels_list:
        if label == "safe" or label in CATEGORY_EXPLANATIONS:
            continue

        matched = False

        for group, keywords in LEGACY_KEYWORDS.items():
            if group == "safe" or group == "non_violent":
                continue

            if any(contains_keyword(label, keyword) for keyword in keywords):
                # Only use this compatibility rule for legacy labels.
                legacy_explanations = {
                    "discrimination": (
                        "The text may contain discriminatory, stereotypical, "
                        "or unjust language toward a person or group."
                    ),
                    "hate_speech": (
                        "The text contains offensive or potentially hateful language."
                    ),
                    "threat": (
                        "The text appears to contain a threat or an intention "
                        "to cause harm."
                    ),
                    "violence": (
                        "The text contains references to or encouragement of violence."
                    ),
                    "harassment": (
                        "The text appears to demean, insult, intimidate, "
                        "or target another person."
                    ),
                    "abuse": (
                        "The text contains abusive or degrading language."
                    ),
                    "toxicity": (
                        "The text contains toxic or harmful language."
                    ),
                    "sexual_content": (
                        "The text contains sexual or potentially explicit content."
                    ),
                    "self_harm": (
                        "The text contains content related to self-harm or suicide."
                    ),
                    "privacy": (
                        "The text may involve sensitive personal or private information."
                    ),
                    "prompt_injection": (
                        "The text appears to contain instructions designed to "
                        "manipulate or bypass an AI system's intended behavior."
                    ),
                    "cyber_threat": (
                        "The text may involve malicious or unauthorized cyber activity."
                    ),
                    "illegal_activity": (
                        "The text may involve or encourage potentially illegal activity."
                    )
                }

                if group in legacy_explanations:
                    explanation = legacy_explanations[group]
                    if explanation not in explanations:
                        explanations.append(explanation)

                matched = True
                break

        if not matched:
            explanations.append(
                f"The classifier predicted '{label}', but no explanation "
                "has been mapped to this category yet."
            )

    # Handle Safe only when the classifier explicitly returns Safe.
    if "safe" in labels_list:
        if len(labels_list) == 1:
            explanations.append(
                "No safety-related category was detected by the classifier."
            )
        else:
            explanations.append(
                "The classifier returned Safe alongside other labels; "
                "review the predictions for consistency."
            )

    # Empty predictions are not automatically proof of safety.
    if not labels_list:
        explanations.append(
            "The classifier returned no category labels. "
            "This does not independently guarantee that the content is safe."
        )

    return explanations
```

This cell defines the `explain_prediction` function, which takes a list of labels (predicted categories) and generates human-readable explanations based on predefined rules. It checks for keywords within the labels (e.g., 'discrimination', 'hate_speech', 'violence') and appends corresponding explanatory phrases to a list. If the model returns labels for which no explanation rule is mapped, it provides a generic message. If no labels are detected at all, it indicates that no safety-related category was found.

```python
class ImageSafetyClassifier(nn.Module):

    def __init__(self, input_dim=768, num_classes=5):
        super().__init__()

        self.classifier = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Dropout(0.3),

            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.2),

            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        return self.classifier(x)
```

```python
from tqdm.auto import tqdm
import torch
import numpy as np

def extract_vit_features(dataset_split, indices, batch_size=16):

    features = []

    for start in tqdm(
        range(0, len(indices), batch_size),
        desc="Extracting ViT features"
    ):

        batch_indices = indices[start:start + batch_size]

        # Get images
        images = [
            dataset_split[i]["image"].convert("RGB")
            for i in batch_indices
        ]

        # Convert images into ViT input format
        inputs = processor(
            images=images,
            return_tensors="pt"
        )

        # Move to GPU/CPU
        inputs = {
            key: value.to(device)
            for key, value in inputs.items()
        }

        # Extract features
        with torch.no_grad():
            outputs = vit(**inputs)

        # CLS token = overall image representation
        batch_features = outputs.last_hidden_state[:, 0, :]

        # Move from GPU to CPU
        features.append(batch_features.cpu().numpy())

    return np.concatenate(features, axis=0)
```

```python
#IMAGE TESTING

from google.colab import files
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import torch


def analyze_image(image):

    # =====================================================
    # 1. VISION TRANSFORMER BRANCH
    # =====================================================

    vit_inputs = processor(
        images=image,
        return_tensors="pt"
    )

    vit_inputs = {
        key: value.to(device)
        for key, value in vit_inputs.items()
    }

    with torch.no_grad():
        vit_outputs = vit(**vit_inputs)

    # CLS token = 768-dimensional image representation
    visual_feature = vit_outputs.last_hidden_state[:, 0, :]

    visual_feature_np = visual_feature.cpu().numpy()

    # Image classifier prediction
    vit_prediction = image_classifier.predict(
        visual_feature_np
    )[0]

    vit_probabilities = image_classifier.predict_proba(
        visual_feature_np
    )[0]

    vit_label = "Unsafe" if vit_prediction == 1 else "Safe"


    # =====================================================
    # 2. OCR BRANCH
    # =====================================================

    image_np = np.array(image)

    ocr_results = ocr_reader.readtext(image_np)

    extracted_text_parts = []

    for result in ocr_results:
        _, text, confidence = result

        # Keep reasonably confident OCR text
        if confidence >= 0.30:
            extracted_text_parts.append(text)

    extracted_text = " ".join(extracted_text_parts).strip()


    # Send OCR text through our ORIGINAL text model
    if extracted_text:

        ocr_labels = predict_text(extracted_text)

        # Anything other than Safe means OCR found
        # a safety category
        ocr_unsafe = any(
            str(label).lower() != "safe"
            for label in ocr_labels
        )

    else:

        ocr_labels = ["Safe"]
        ocr_unsafe = False


    # =====================================================
    # 3. FINAL IMAGE DECISION
    # =====================================================

    final_unsafe = (
        vit_prediction == 1
        or ocr_unsafe
    )

    final_label = "Unsafe" if final_unsafe else "Safe"


    # =====================================================
    # 4. DISPLAY RESULTS
    # =====================================================

    print("=" * 55)
    print("              IMAGE SAFETY ANALYSIS")
    print("=" * 55)

    print("\n[VISUAL BRANCH - ViT]")
    print("-----------------------")
    print("Prediction       :", vit_label)
    print("Safe probability :", f"{vit_probabilities[0]:.4f}")
    print("Unsafe probability:", f"{vit_probabilities[1]:.4f}")

    print("\n[OCR BRANCH]")
    print("-----------------------")

    if extracted_text:
        print("Extracted text:")
        print(extracted_text)

        print("\nText safety categories:")
        for label in ocr_labels:
            print("-", label)
    else:
        print("No text detected.")

    print("\n[FINAL DECISION]")
    print("-----------------------")
    print("Final result:", final_label)

    print("=" * 55)

    return {
        "final_label": final_label,
        "vit_label": vit_label,
        "vit_safe_probability": float(vit_probabilities[0]),
        "vit_unsafe_probability": float(vit_probabilities[1]),
        "ocr_text": extracted_text,
        "ocr_labels": ocr_labels
    }


# =========================================================
# UPLOAD IMAGE
# =========================================================

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

image = Image.open(image_path).convert("RGB")


# Display image
plt.figure(figsize=(7, 7))
plt.imshow(image)
plt.axis("off")
plt.title("Input Image")
plt.show()


# =========================================================
# RUN COMPLETE PIPELINE
# =========================================================

result = analyze_image(image)
```

```python
# ===== NEW CELL A: Groq VLM contextual assessment (standalone) =====
!pip -q install groq

import os, io, json, re, time, base64
from PIL import Image
import groq
from groq import Groq

try:
    from google.colab import userdata
    _GROQ_KEY = userdata.get("GROQ_API_KEY")
except Exception:
    _GROQ_KEY = os.environ.get("GROQ_API_KEY")
if not _GROQ_KEY:
    raise RuntimeError("Set GROQ_API_KEY in Colab Secrets or as an environment variable.")

groq_client = Groq(api_key=_GROQ_KEY)
GROQ_VLM_MODEL = "qwen/qwen3.8-27b"   # verify against console.groq.com/docs/vision

VLM_SYSTEM_PROMPT = """You are an image safety reviewer. Judge whether the image is UNSAFE
(e.g. graphic violence, sexual content, hate, harassment, self-harm, illegal activity,
shocking content) or SAFE, using the visible content AND its context.

Rules:
- Base your answer only on what is actually visible. Do not invent text or details.
- Do not mark an image Unsafe only because a sensitive word appears. Distinguish real
  harm from news reporting, education, counterspeech, quotation, or harmless use.
- OCR text is provided as hints and may be wrong. Trust the image over the OCR text.
- If the meaning is unclear, say so through the uncertainty field.

Respond with ONLY a JSON object with exactly these keys:
{"label": "Safe" or "Unsafe",
 "reason": "one or two sentences",
 "visible_text": "text you can read in the image, or empty string",
 "evidence": ["short visual or textual details supporting the label"],
 "uncertainty": "low" or "medium" or "high"}"""

def _image_to_data_url(image, max_side=1024):
    img = image.convert("RGB")
    img.thumbnail((max_side, max_side))
    buf = io.BytesIO()
    img.save(buf, format="JPEG", quality=90)
    return "data:image/jpeg;base64," + base64.b64encode(buf.getvalue()).decode()

def _parse_vlm_json(text):
    text = re.sub(r"<think>.*?</think>", "", text or "", flags=re.S).strip()
    m = re.search(r"\{.*\}", text, flags=re.S)
    if not m:
        raise ValueError("no JSON object found")
    return json.loads(m.group(0))

def _validate_vlm(d):
    label = str(d.get("label", "")).strip().capitalize()
    if label not in ("Safe", "Unsafe"):
        raise ValueError(f"invalid label: {d.get('label')!r}")
    unc = str(d.get("uncertainty", "")).strip().lower()
    if unc not in ("low", "medium", "high"):
        unc = "unknown"
    ev = d.get("evidence", [])
    if isinstance(ev, str):
        ev = [ev]
    return {"label": label,
            "reason": str(d.get("reason", "")),
            "visible_text": str(d.get("visible_text", "")),
            "evidence": [str(e) for e in ev][:6],
            "uncertainty": unc}

def groq_vlm_assess(image, ocr_text="", max_retries=3):
    """Returns {'status': 'ok'|'error', 'result': {...}|None, 'error': str|None}."""
    try:
        data_url = _image_to_data_url(image)
    except Exception as e:
        return {"status": "error", "result": None, "error": f"unsupported image: {e}"}

    hint = ocr_text.strip() if ocr_text and ocr_text.strip() else "(no text detected by OCR)"
    messages = [
        {"role": "system", "content": VLM_SYSTEM_PROMPT},
        {"role": "user", "content": [
            {"type": "text", "text": f"OCR hint (may contain errors):\n{hint}"},
            {"type": "image_url", "image_url": {"url": data_url}},
        ]},
    ]
    last_err = None
    for attempt in range(max_retries):
        try:
            resp = groq_client.chat.completions.create(
                model=GROQ_VLM_MODEL, messages=messages,
                temperature=0, max_completion_tokens=1024,
                response_format={"type": "json_object"},
            )
            parsed = _validate_vlm(_parse_vlm_json(resp.choices[0].message.content))
            return {"status": "ok", "result": parsed, "error": None}
        except groq.RateLimitError as e:
            last_err = f"rate limit: {e}"; time.sleep(2 ** (attempt + 2))
        except (groq.APIConnectionError, groq.InternalServerError) as e:
            last_err = f"api/connection: {e}"; time.sleep(2 ** attempt)
        except groq.APIStatusError as e:
            return {"status": "error", "result": None, "error": f"api status {e.status_code}: {e}"}
        except (ValueError, json.JSONDecodeError) as e:
            last_err = f"malformed response: {e}"   # retry once or twice
        except Exception as e:
            return {"status": "error", "result": None, "error": f"unexpected: {e}"}
    return {"status": "error", "result": None, "error": last_err}

# --- quick independent test (replace with your own image paths) ---
test_paths = []   # e.g. ["/content/meme1.jpg", "/content/poster.png"]
for p in test_paths:
    out = groq_vlm_assess(Image.open(p))
    print(p, "\n", json.dumps(out, indent=2), "\n")
```

```python

# ===== NEW CELL: Combined ViT + OCR + Groq image safety =====

import contextlib
import io
from PIL import Image

def analyze_image_with_groq(image):
    """
    Run the existing ViT + OCR pipeline and Groq VLM.
    Fusion rule: Unsafe if either successful branch says Unsafe.
    If Groq fails, fall back to the existing pipeline.
    """

    # Normalize image to RGB PIL format
    if not isinstance(image, Image.Image):
        image = Image.fromarray(image)
    image = image.convert("RGB")

    # 1. Existing ViT + OCR pipeline (no retraining)
    with contextlib.redirect_stdout(io.StringIO()):
        baseline = analyze_image(image)

    if not isinstance(baseline, dict):
        raise TypeError(
            "analyze_image() must return a dictionary."
        )

    baseline_label = str(
        baseline.get("final_label", "")
    ).strip().capitalize()

    if baseline_label not in ("Safe", "Unsafe"):
        raise ValueError(
            f"Unexpected baseline label: {baseline_label!r}"
        )

    # Find OCR text if the existing function exposes it
    ocr_text = ""
    for key in ("ocr_text", "extracted_text", "visible_text", "text"):
        value = baseline.get(key)
        if isinstance(value, str) and value.strip():
            ocr_text = value.strip()
            break

    # 2. Groq VLM contextual assessment
    groq_output = groq_vlm_assess(
        image,
        ocr_text=ocr_text
    )

    groq_result = groq_output.get("result") or {}
    groq_status = groq_output.get("status", "error")
    groq_label = groq_result.get("label")

    # 3. Fuse predictions; never let an API failure erase baseline output
    if groq_status == "ok" and groq_label in ("Safe", "Unsafe"):
        final_label = (
            "Unsafe"
            if baseline_label == "Unsafe" or groq_label == "Unsafe"
            else "Safe"
        )
        fusion_rule = "OR: baseline Unsafe OR Groq Unsafe"
    else:
        final_label = baseline_label
        fusion_rule = "Baseline fallback: Groq unavailable or invalid"

    # 4. Display transparent branch outputs
    print("=" * 55)
    print("       COMBINED IMAGE SAFETY ANALYSIS")
    print("=" * 55)

    print("\n[EXISTING ViT + OCR PIPELINE]")
    print("Baseline prediction:", baseline_label)

    print("\n[GROQ VISION-LANGUAGE MODEL]")
    print("API status:", groq_status)

    if groq_status == "ok" and groq_label in ("Safe", "Unsafe"):
        print("VLM prediction:", groq_label)
        print("Reason:", groq_result.get("reason", "Not provided"))
        print("Evidence:", groq_result.get("evidence", []))
        print("Uncertainty:", groq_result.get("uncertainty", "unknown"))
    else:
        print("Groq error:", groq_output.get("error"))

    print("\n[FUSION]")
    print("Rule:", fusion_rule)
    print("FINAL RESULT:", final_label)
    print("=" * 55)

    return {
        "final_label": final_label,
        "baseline_label": baseline_label,
        "groq_status": groq_status,
        "groq_label": groq_label,
        "groq_reason": groq_result.get("reason"),
        "groq_evidence": groq_result.get("evidence", []),
        "groq_uncertainty": groq_result.get("uncertainty"),
        "ocr_text": ocr_text,
        "fusion_rule": fusion_rule,
        "groq_error": groq_output.get("error")
    }

print("Combined image-analysis function is ready.")
```

```python

# STEP 17: Compare ViT, OCR alone, and ViT + OCR

import ast
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

df_vit = pd.read_csv("/content/heldout_vit_results.csv")
df_fusion = pd.read_csv("/content/heldout_vit_ocr_fusion_results.csv")

# Join by original test index to ensure identical examples
df_eval = df_vit.merge(
    df_fusion[
        ["image_index", "ocr_labels", "fusion_label"]
    ],
    on="image_index",
    how="inner",
    validate="one_to_one"
)

def parse_labels(value):
    if isinstance(value, list):
        return value
    try:
        return ast.literal_eval(value)
    except (ValueError, SyntaxError):
        return []

# Reconstruct OCR-only decision using the existing fusion rule
df_eval["ocr_unsafe"] = df_eval["ocr_labels"].apply(
    lambda labels: any(
        str(label).lower() != "safe"
        for label in parse_labels(labels)
    )
)

df_eval["ocr_label"] = df_eval["ocr_unsafe"].map({
    True: "Unsafe",
    False: "Safe"
})

def calculate_metrics(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Unsafe Precision": precision_score(
            y_true, y_pred, pos_label="Unsafe", zero_division=0
        ),
        "Unsafe Recall": recall_score(
            y_true, y_pred, pos_label="Unsafe", zero_division=0
        ),
        "Unsafe F1": f1_score(
            y_true, y_pred, pos_label="Unsafe", zero_division=0
        ),
        "Macro F1": f1_score(
            y_true, y_pred, average="macro", zero_division=0
        )
    }

y_true = df_eval["y_true"]

summary = pd.DataFrame({
    "ViT alone": calculate_metrics(y_true, df_eval["vit_label"]),
    "OCR alone": calculate_metrics(y_true, df_eval["ocr_label"]),
    "ViT + OCR": calculate_metrics(y_true, df_eval["fusion_label"])
}).T

print("Examples compared:", len(df_eval))
print("\nMetrics (percent):")
print((summary * 100).round(2).to_string())

for branch, column in [
    ("ViT alone", "vit_label"),
    ("OCR alone", "ocr_label"),
    ("ViT + OCR", "fusion_label")
]:
    print(f"\n{branch} confusion matrix (rows=true, columns=predicted)")
    print(confusion_matrix(
        y_true,
        df_eval[column],
        labels=["Safe", "Unsafe"]
    ))

summary.to_csv("/content/final_image_branch_comparison.csv")
df_eval.to_csv("/content/final_image_evaluation_details.csv", index=False)

print("\nSaved:")
print("/content/final_image_branch_comparison.csv")
print("/content/final_image_evaluation_details.csv")
```
